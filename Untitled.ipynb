```python
import py4hw

class FullAdder(py4hw.Logic):
    def __init__(self, parent, name, x, y, ci, s, co):
        super().__init__(parent, name)
        
        x = self.addIn('x', x)
        y = self.addIn('y', y)
        ci = self.addIn('ci', ci)
        s = self.addOut('s', s)
        co = self.addOut('co', co)
        
        w1 = self.wire('w1', 1)
        w2 = self.wire('w2', 1)
        w3 = self.wire('w3', 1)
        
        
        py4hw.Xor2(self, 'g1', x, y, w1)
        py4hw.Xor2(self, 'g2', w1, ci, s)
        py4hw.And2(self, 'g3', w1, ci, w2)
        py4hw.And2(self, 'g4', x, y, w3)
        py4hw.Or2(self, 'g5', w2, w3, co)
        
```

```python
hw = py4hw.HWSystem()
x = hw.wire('a', 1)
y = hw.wire('b', 1)

ci = hw.wire('ci', 1)
co = hw.wire('co', 1)
s = hw.wire('s', 1)

fa = FullAdder(hw, 'fa', x, y, ci, s, co)
sch = py4hw.Schematic(fa)
sch.draw()
```

```python
hw = py4hw.HWSystem()
x = hw.wire('a', 1)
y = hw.wire('b', 1)
ci = hw.wire('ci', 1)
co = hw.wire('co', 1)
s = hw.wire('s', 1)

py4hw.Sequence(hw, 'x', [0, 1], x)
py4hw.Sequence(hw, 'y', [0, 0, 1, 1], y)
py4hw.Sequence(hw, 'ci', [0, 0, 0, 0, 1, 1, 1, 1], ci)

fa = FullAdder(hw, 'fa', x, y, ci, s, co)
wvf = py4hw.Waveform(hw, 'wvf', [x, y, ci, s, co])
hw.getSimulator().clk(16)
wvf.draw_wavedrom()
```

```python
import py4hw

class NBitAdder(py4hw.Logic):
    def __init__(self, parent, name, a, b, ci, s, co):
        super().__init__(parent, name)
        
        a = self.addIn('a', a)
        b = self.addIn('b', b)
        ci = self.addIn('ci', ci)
        s = self.addOut('s', s)
        co = self.addOut('co', co)
        
        width = a.getWidth()
        
        carry_wires = [ci] + [self.wire(f'c_{i}', 1) for i in range(width - 1)] + [co]
        sum_wires = [self.wire(f's_{i}', 1) for i in range(width)]
        
        for i in range(width):
            a_bit = self.wire(f'a_{i}', 1)
            b_bit = self.wire(f'b_{i}', 1)
            
            py4hw.Bit(self, f'bit_a_{i}', a, i, a_bit)
            py4hw.Bit(self, f'bit_b_{i}', b, i, b_bit)
            
            FullAdder(self, f'fa_{i}', a_bit, b_bit, carry_wires[i], sum_wires[i], carry_wires[i+1])
            
        py4hw.ConcatenateLSBF(self, 'concat_sum', sum_wires, s)
```

```python
hw = py4hw.HWSystem()

# Create 4-bit input wires and 1-bit carry signals
a = hw.wire('a', 4)
b = hw.wire('b', 4)
ci = hw.wire('ci', 1)
co = hw.wire('co', 1)
s = hw.wire('s', 4)

# Instantiate a 4-bit adder
adder = NBitAdder(hw, 'adder4', a, b, ci, s, co)

# Draw schematic
sch = py4hw.Schematic(adder)
sch.draw()
```

```python
import numpy as np

# Define RandomUniform generator
class RandomUniform(py4hw.Logic):
    def __init__(self, parent, name, r):
        super().__init__(parent, name)
        self.r = self.addOut('r', r)
        
    def clock(self):
        w = self.r.getWidth()
        v = int(np.random.randint(0, 1 << w))
        self.r.prepare(v)

# Set up simulation
hw = py4hw.HWSystem()

a = hw.wire('a', 4)
b = hw.wire('b', 4)
ci = hw.wire('ci', 1)
co = hw.wire('co', 1)
s = hw.wire('s', 4)

# Drive inputs with random numbers
RandomUniform(hw, 'rand_a', a)
RandomUniform(hw, 'rand_b', b)
py4hw.Sequence(hw, 'ci_seq', [0], ci)

# Instantiate adder and waveform recorder
NBitAdder(hw, 'adder', a, b, ci, s, co)
wvf = py4hw.Waveform(hw, 'wvf', [a, b, ci, s, co])

# Run simulation for 8 clock cycles
hw.getSimulator().clk(8)
wvf.draw_wavedrom()
```
